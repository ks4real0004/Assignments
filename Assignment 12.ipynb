```python
# PLANT DISEASE CLASSIFICATION USING CNN
# Kaggle - Complete Code

import os
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report

# 1. Dataset path
BASE = "/kaggle/input/datasets/vipoooool/new-plant-diseases-dataset"

TRAIN = VALID = None

for root, dirs, files in os.walk(BASE):
    for d in dirs:
        if d.lower() == "train":
            TRAIN = os.path.join(root, d)
        elif d.lower() in ("valid", "validation"):
            VALID = os.path.join(root, d)

if TRAIN is None:
    raise FileNotFoundError("Train folder not found!")

# 2. Settings
SIZE = (96, 96)
BATCH = 32
EPOCHS = 5
TRAIN_STEPS = 20
VAL_STEPS = 5

# 3. Load dataset
train = tf.keras.utils.image_dataset_from_directory(
    TRAIN,
    image_size=SIZE,
    batch_size=BATCH,
    shuffle=True,
    seed=123
)

classes = train.class_names
num_classes = len(classes)

if VALID:
    val = tf.keras.utils.image_dataset_from_directory(
        VALID,
        class_names=classes,
        image_size=SIZE,
        batch_size=BATCH,
        shuffle=False
    )
else:
    train = tf.keras.utils.image_dataset_from_directory(
        TRAIN,
        validation_split=0.2,
        subset="training",
        seed=123,
        image_size=SIZE,
        batch_size=BATCH
    )
    val = tf.keras.utils.image_dataset_from_directory(
        TRAIN,
        validation_split=0.2,
        subset="validation",
        seed=123,
        image_size=SIZE,
        batch_size=BATCH
    )
    classes = train.class_names
    num_classes = len(classes)

print("\nNUMBER OF CLASSES:", num_classes)
print("CLASS NAMES:", classes)

# 4. Display 3 plant images with TRUE labels
images, labels = next(iter(train))

plt.figure(figsize=(12, 4))

for i in range(min(3, len(images))):
    plt.subplot(1, 3, i + 1)
    plt.imshow(images[i].numpy().astype("uint8"))
    plt.title("True label:\n" + classes[int(labels[i])])
    plt.axis("off")

plt.suptitle("Sample Plant Images and True Labels")
plt.tight_layout()
plt.show()

# 5. Prepare dataset
train = train.prefetch(tf.data.AUTOTUNE)
val = val.prefetch(tf.data.AUTOTUNE)

# 6. Build CNN
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(96, 96, 3)),
    tf.keras.layers.Rescaling(1./255),

    tf.keras.layers.Conv2D(16, 3, activation="relu"),
    tf.keras.layers.MaxPooling2D(),

    tf.keras.layers.Conv2D(32, 3, activation="relu"),
    tf.keras.layers.MaxPooling2D(),

    tf.keras.layers.GlobalAveragePooling2D(),
    tf.keras.layers.Dense(64, activation="relu"),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(num_classes, activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# 7. Model summary
print("\nCNN MODEL SUMMARY:")
model.summary()

# 8. Train CNN
history = model.fit(
    train,
    validation_data=val,
    epochs=EPOCHS,
    steps_per_epoch=TRAIN_STEPS,
    validation_steps=VAL_STEPS
)

# 9. Evaluate
loss, accuracy = model.evaluate(
    val, steps=VAL_STEPS, verbose=0
)

print(f"\nValidation accuracy: {accuracy * 100:.2f}%")
print(f"Validation loss: {loss:.4f}")

# 10. Accuracy and loss graphs
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(history.history["accuracy"], label="Training")
plt.plot(history.history["val_accuracy"], label="Validation")
plt.title("CNN Accuracy")
plt.xlabel("Epoch")
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history.history["loss"], label="Training")
plt.plot(history.history["val_loss"], label="Validation")
plt.title("CNN Loss")
plt.xlabel("Epoch")
plt.legend()

plt.tight_layout()
plt.show()

# 11. Display 3 images with TRUE and PREDICTED labels
images, labels = next(iter(val))
predictions = model.predict(images, verbose=0)

plt.figure(figsize=(13, 4))

for i in range(min(3, len(images))):
    true_label = classes[int(labels[i])]
    predicted_label = classes[int(np.argmax(predictions[i]))]

    plt.subplot(1, 3, i + 1)
    plt.imshow(images[i].numpy().astype("uint8"))
    plt.title(
        f"True: {true_label}\nPredicted: {predicted_label}",
        fontsize=8
    )
    plt.axis("off")

plt.suptitle("Actual vs Predicted Plant Disease")
plt.tight_layout()
plt.show()

# 12. Classification report on limited validation batches
y_true, y_pred = [], []

for imgs, labs in val.take(VAL_STEPS):
    probs = model.predict(imgs, verbose=0)
    y_true.extend(labs.numpy())
    y_pred.extend(np.argmax(probs, axis=1))

print("\nCLASSIFICATION REPORT:")
print(classification_report(
    y_true,
    y_pred,
    labels=list(range(num_classes)),
    target_names=classes,
    zero_division=0
))

# 13. Save model
model.save("/kaggle/working/plant_disease_cnn.keras")
print("\nModel saved successfully!")

# 14. Predict any new plant image
def predict_disease(image_path):
    img = tf.keras.utils.load_img(image_path, target_size=SIZE)
    arr = tf.keras.utils.img_to_array(img)
    arr = np.expand_dims(arr, axis=0)

    probs = model.predict(arr, verbose=0)[0]
    idx = int(np.argmax(probs))

    print("Predicted disease:", classes[idx])
    print(f"Confidence: {probs[idx] * 100:.2f}%")

    plt.imshow(img)
    plt.title("Predicted: " + classes[idx])
    plt.axis("off")
    plt.show()

# Example:
# predict_disease("/kaggle/working/test_leaf.jpg")
```
